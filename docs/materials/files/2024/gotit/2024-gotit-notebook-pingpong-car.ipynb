```python
# 핑퐁 로봇 제어 모듈 불러오기
from pingpongthread import PingPongThread
```

```python
# '기다리기' 기능을 사용하기 위한 time 모듈 가져오기
import time
```

```python
# 키보드를 사용하기 위한 keyboard 모듈 가져오기
import keyboard
```

```python
# 2개 로봇을 연결하는 인스턴스 생성
PingPong = PingPongThread(number=2)
```

```python
# 로봇 제어 시작
PingPong.start()
PingPong.wait_until_full_connect()
```

```python
# 전진하기
PingPong.run_motor([1, 2], [-15, 15]) # 모터가 반대 방향에 있으므로 하나는 마이너스임
time.sleep(2) # 2초간 전진
PingPong.run_motor([1, 2], 'stop')
```

```python
# 우회전
PingPong.run_motor([1, 2], [-7, -7])
time.sleep(0.5)
PingPong.run_motor([1, 2], 'stop')
```

```python
# 좌회전
PingPong.run_motor([1, 2], [7, 7])
time.sleep(0.5)
PingPong.run_motor([1, 2], 'stop')
```

```python
# 키보드로 전진하기
while not keyboard.is_pressed("esc") :
    if keyboard.is_pressed("up") :
        PingPong.run_motor([1, 2], [-15, 15])
    elif keyboard.is_pressed("space") :
         PingPong.run_motor([1, 2], 'stop')
```

```python
# 키보드로 핑퐁 오토카 조종하기
while not keyboard.is_pressed("esc") :
    if keyboard.is_pressed("up") :
        PingPong.run_motor([1, 2], [-15, 15])
    elif keyboard.is_pressed("right") :
        PingPong.run_motor([1, 2], [-7, -7])
    elif keyboard.is_pressed("left") :
        PingPong.run_motor([1, 2], [7, 7])
    elif keyboard.is_pressed("down") :
        PingPong.run_motor([1, 2], [15, -15])
      elif keyboard.is_pressed("space") :
                PingPong.run_motor([1, 2], 'stop ')
```

```python
# 웹캠 열기
PingPong.webcam_open(0)
```

```python
# 첫 번째 스냅샷 찍기
PingPong.webcam_take_snapshots("car/ahead")
```

```python
# 두 번째 스냅샷 찍기
PingPong.webcam_take_snapshots("car/stop")
```

```python
# 첫 번째 이미지 클래스 만들기
go = PingPongThread.ImageClass("go", "car/ahead")
```

```python
# 두 번째 이미지 클래스 만들기
stop = PingPongThread.ImageClass("stop", "car/stop")
```

```python
# 이미지 클래스 학습하기
trained = PingPong.train_classes("model/car.json", 5, 1, go, stop)
```

```python
# 모션으로 핑퐁 오토카 제어하기
tool = PingPongThread.FramesPredictor(trained, 3)
while not keyboard.is_pressed('esc'):
    PingPong.clear_output()
    frame = PingPong.webcam_get_frame(window="GET_frame")
    result = tool.image_predict_and_accum(frame)
    print(result)
    time.sleep(0.1)
    max_class = max(result, key=result.get)
    print(max_class)
    if max_class == 'go' :
        PingPong.run_motor([1, 2], [-15, 15])
    elif max_class == 'stop' :
        PingPong.run_motor([1, 2], 'stop')
```

```python
# 웹캠 종료하기
PingPong.webcam_close()
```

```python
# 로봇 제어 스레드 종료
PingPong.end()
```
