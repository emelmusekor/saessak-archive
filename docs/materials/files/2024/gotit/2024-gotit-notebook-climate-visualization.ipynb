# 데이터 시각화 코드

```python
import pandas as pd
import matplotlib.pyplot as plt
import matplotlib.font_manager as fm

while True:
    user_input = input("1: 평균기온(℃), 2: 평균최저기온(℃), 3: 평균최고기온(℃) 중 하나를 입력하세요: ")
    if user_input == '1':
        column_name = '평균기온(℃)'
        break
    elif user_input == '2':
        column_name = '평균최저기온(℃)'
        break
    elif user_input == '3':
        column_name = '평균최고기온(℃)'
        break
    else:
        print("잘못된 입력입니다. 다시 입력해주세요.")

mydata = pd.read_csv("/content/temp_data.csv", encoding="cp949")
mydata = mydata[['년', column_name]]

plt.figure(figsize=(15, 8))
fe = fm.FontEntry(fname=r'/usr/share/fonts/truetype/nanum/NanumGothic.ttf', name='NanumGothic')
fm.fontManager.ttflist.insert(0, fe)
plt.rcParams.update({'font.size': 18, 'font.family': 'NanumGothic'})
ax = mydata.set_index('년')[column_name].plot(kind='line', marker='s')
ax.set_ylabel(column_name)
ax.set_xlabel("연도")
plt.title(column_name + ' 변화 그래프')
plt.show()
```
