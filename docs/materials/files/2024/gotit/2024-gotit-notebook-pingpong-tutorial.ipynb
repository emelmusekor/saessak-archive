```python
# 핑퐁 센서 수신 설정
PingPong.receive_sensor_data(1, method="periodic", period=0.5)
```

```python
# 버튼 눌림 출력하기
while True:
    time.sleep(0.5)   # 0.5초 기다리기
    if PingPong.get_current_button(1) == 1:
        print("버튼 눌림")
        break
```

```python
# 버튼으로 모터 제어하기
PingPong.run_motor(1, 30)
while True:
    time.sleep(0.5)
    if PingPong.get_current_button(1) == 1 :
        PingPong.run_motor(1, 'stop')
        break
```

```python
# 근접 센서 값 출력하기
while not PingPong.get_current_button(1) == 1:
    print(PingPong.get_current_proxy(1))    # 근접 센서 값 출력
    time.sleep(0.5)
```

```python
# 장애물 감지 후 정지
PingPong.run_motor(1,-30)
while True:
    print(PingPong.get_current_proxy(1))
    if PingPong.get_current_proxy(1)>100:
        print("멈춤!")
        PingPong.run_motor(1, 'stop')
        break
    time.sleep(0.5)
```

```python
# 센서 수신 주기 재설정
PingPong.receive_sensor_data(1, method="periodic", period=0.1)
```

```python
# 기울기 센서 값 출력하기
while not PingPong.get_current_button(1) == 1:
    gyro_value = PingPong.get_current_gyro(1)
    print("roll: ", gyro_value[0])
    print("pitch: ", gyro_value[1])
    print("yaw: ", gyro_value[2])
    time.sleep(0.5)
```

```python
# roll 값을 이용한 모터 회전
while not PingPong.get_current_button(1) == 1:
    gyro_value = PingPong.get_current_gyro(1)
    print(gyro_value[0])
    if gyro_value[0] < -50:
        print("네모 방향으로 회전!")
        PingPong.run_motor(1, 30)
    elif gyro_value[0] > 50:
        print("별 방향으로 회전!")
        PingPong.run_motor(1, -30)
    time.sleep(0.1)
PingPong.run_motor(1,0)  # 버튼 누르면 멈추기
```

```python
# 센서 수신 종료
PingPong.stop_sensor_data(1)
```

```python
# 로봇 제어 스레드 종료
PingPong.end()
```
